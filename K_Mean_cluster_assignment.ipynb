PCA Step by Step (From Scratch!)
Step 1: Center the Data

```python
X_centered = X - X.mean(axis=0)
print(f"Mean after centering (should be ~0): {X_centered.mean(axis=0).round(10)}")
```

### Step 2: Calculate the Covariance Matrix

The covariance matrix describes how much the different features in the dataset vary together. For PCA, we need the covariance matrix of the centered data.

```python
import numpy as np
covariance_matrix = np.cov(X_centered, rowvar=False)

print("Covariance Matrix (first 5x5 block):\n", covariance_matrix[:5, :5])
print("Shape of Covariance Matrix:", covariance_matrix.shape)
```

### Step 3: Calculate Eigenvalues and Eigenvectors

Eigenvalues and eigenvectors are crucial for PCA. Eigenvectors represent the principal components (new axes), and eigenvalues represent the magnitude of variance along those axes.

```python
import numpy as np
eigenvalues, eigenvectors = np.linalg.eigh(covariance_matrix)

# Sort eigenvalues and eigenvectors in descending order
sorted_indices = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[sorted_indices]
eigenvectors = eigenvectors[:, sorted_indices]
```

### Step 4: Choose the Number of Principal Components

We can determine the number of principal components to keep by looking at the explained variance ratio. This tells us how much variance each principal component captures.

```python
import numpy as np
import matplotlib.pyplot as plt

explained_variance_ratio = eigenvalues / np.sum(eigenvalues)
cumulative_explained_variance = np.cumsum(explained_variance_ratio)

plt.figure(figsize=(10, 6))
plt.plot(range(1, len(explained_variance_ratio) + 1), cumulative_explained_variance, marker='o', linestyle='--')
plt.title('Cumulative Explained Variance by Principal Component')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.grid(True)
plt.show()

# Let's find how many components explain at least 95% of the variance
num_components_95_var = np.where(cumulative_explained_variance >= 0.95)[0][0] + 1
print(f"Number of components to explain 95% variance: {num_components_95_var}")
```

### Step 5: Project Data onto Principal Components

Now, we project the centered data onto the chosen principal components to reduce its dimensionality.

```python
import numpy as np
# We'll use the number of components that explain 95% of the variance
selected_eigenvectors = eigenvectors[:, :num_components_95_var]
X_pca = np.dot(X_centered, selected_eigenvectors)

print("Original data shape:", X.shape)
print("Reduced data shape (PCA):", X_pca.shape)
```

## Finding the best k cluster value for PCA-Reduced Data using the Elbow Method (WCSS)

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss_pca = []

for i in range(1,11):
  model_pca = KMeans(n_clusters = i, random_state=42, n_init=10)
  model_pca.fit(X_pca)
  wcss_pca.append(model_pca.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(1,11), wcss_pca, marker="o", linestyle='--')
plt.title('Elbow Method for Optimal K on PCA-Reduced Data')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Within Cluster Sum of Squares)')
plt.grid(True)
plt.show()
```

## Finding the best k cluster value for PCA-Reduced Data using the Silhouette Score

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

silhouette_scores_pca = []

# We need at least 2 clusters for silhouette score
for i in range(2, 11):
  model_pca = KMeans(n_clusters = i, random_state=42, n_init=10)
  cluster_labels = model_pca.fit_predict(X_pca)
  score = silhouette_score(X_pca, cluster_labels)
  silhouette_scores_pca.append(score)

plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), silhouette_scores_pca, marker="o", linestyle='--')
plt.title('Silhouette Score for Optimal K on PCA-Reduced Data')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()
```

### Outlier Detection and Removal using IQR Method

Outliers are data points that significantly deviate from other observations. The Interquartile Range (IQR) method is a common technique to identify them. An outlier is typically defined as a data point that falls below Q1 - 1.5 * IQR or above Q3 + 1.5 * IQR, where Q1 is the first quartile, Q3 is the third quartile, and IQR = Q3 - Q1.

```python
import numpy as np

# Let's consider 'BALANCE' as an example column for outlier handling
# First, let's create a copy of the PCA-reduced data to work with
# and convert it back to a DataFrame for easier column handling if needed
df_pca = pd.DataFrame(X_pca, columns=[f'PC_{i+1}' for i in range(X_pca.shape[1])])

original_shape = df_pca.shape
print(f"Original shape of PCA-reduced data: {original_shape}")

# Calculate Q1, Q3, and IQR for a specific principal component (e.g., PC_1)
Q1 = df_pca['PC_1'].quantile(0.25)
Q3 = df_pca['PC_1'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers_indices = df_pca[(df_pca['PC_1'] < lower_bound) | (df_pca['PC_1'] > upper_bound)].index

print(f"Number of outliers in PC_1 identified: {len(outliers_indices)}")

# Remove outliers
df_pca_no_outliers = df_pca.drop(outliers_indices).reset_index(drop=True)

print(f"Shape after removing outliers from PC_1: {df_pca_no_outliers.shape}")

# You would typically apply this process iteratively or on other relevant columns
# For simplicity, we are showing it for one principal component.
```

### Applying DBSCAN Clustering on PCA-Reduced Data After Outlier Removal

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) is a clustering algorithm that groups together points that are closely packed together, marking as outliers points that lie alone in low-density regions. It requires two main parameters:

*   **`eps` (epsilon):** The maximum distance between two samples for one to be considered as in the neighborhood of the other. If the `eps` value is too small, a large part of the data will not be clustered (will be labeled as noise). If `eps` is too large, it can merge different clusters and points that are far apart might be considered as neighbors.
*   **`min_samples`:** The number of samples (or total weight) in a neighborhood for a point to be considered as a core point. This includes the point itself. A larger `min_samples` value will result in more points being classified as noise.

We will use `X_pca_no_outliers` for DBSCAN.

### Determining Optimal `eps` for DBSCAN using K-distance Graph

To find a suitable value for `eps` in DBSCAN, we can use the K-distance graph. This method involves calculating the distance from each point to its k-th nearest neighbor (where k is usually set to `min_samples`). When these distances are plotted in ascending order, a 'knee' or 'elbow' in the graph often indicates a good `eps` value, representing the point where the density rapidly changes.

We will use `NearestNeighbors` from `sklearn.neighbors` to calculate these distances.

```python
from sklearn.neighbors import NearestNeighbors
import matplotlib.pyplot as plt
import numpy as np

# Let's use a min_samples value, for example, 2 * number of features (12 in X_pca_no_outliers)
k = 2 * X_pca_no_outliers.shape[1] # k = 2 * 12 = 24

neigh = NearestNeighbors(n_neighbors=k)
neigh.fit(X_pca_no_outliers)

distances, indices = neigh.kneighbors(X_pca_no_outliers)

# Sort distances to the k-th nearest neighbor
distances = np.sort(distances[:, k-1], axis=0)

plt.figure(figsize=(12, 6))
plt.plot(distances)
plt.title(f'K-distance Graph for k = {k} Nearest Neighbors')
plt.xlabel('Points sorted by distance to k-th nearest neighbor')
plt.ylabel(f'Distance to {k}-th nearest neighbor')
plt.grid(True)
plt.show()

print(f"Consider looking for a 'knee' in the plot to determine an appropriate `eps` value.")
print(f"A good starting point for `min_samples` could be {k}.")
```

### Re-applying DBSCAN Clustering with Tuned `eps` and `min_samples`

Based on the K-distance graph, we've identified a more suitable `eps` value, and we'll use `min_samples` as `2 * num_features` (24). Let's re-run DBSCAN with these parameters and see the new clustering results.

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
import numpy as np

# Re-initialize and apply DBSCAN with tuned parameters
dbsc_tuned = DBSCAN(eps=1.2, min_samples=24)
clusters_dbsc_tuned = dbsc_tuned.fit_predict(X_pca_no_outliers)

# Number of clusters in labels, ignoring noise if present.
n_clusters_tuned_ = len(set(clusters_dbsc_tuned)) - (1 if -1 in clusters_dbsc_tuned else 0)
n_noise_tuned_ = list(clusters_dbsc_tuned).count(-1)

print(f"Estimated number of clusters (tuned): {n_clusters_tuned_}")
print(f"Estimated number of noise points (tuned): {n_noise_tuned_}")

plt.figure(figsize=(10, 8))

unique_labels_tuned = set(clusters_dbsc_tuned)
colors_tuned = [plt.cm.Spectral(each) for each in np.linspace(0, 1, len(unique_labels_tuned))]

for k, col in zip(unique_labels_tuned, colors_tuned):
    if k == -1: # Noise points
        col = [0, 0, 0, 1] # Black color for noise

    class_member_mask_tuned = (clusters_dbsc_tuned == k)

    xy_tuned = X_pca_no_outliers[class_member_mask_tuned]
    plt.scatter(xy_tuned[:, 0], xy_tuned[:, 1], s=100, c=[col], label=f'Cluster {k}' if k != -1 else 'Noise')

plt.title('DBSCAN Clustering (Tuned) on First Two Principal Components (Outliers Removed)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score

# Calculate the Silhouette Score for the tuned DBSCAN clusters, excluding noise points (-1)
# We need to filter out noise points for silhouette score calculation
valid_points_mask = (clusters_dbsc_tuned != -1)
X_valid = X_pca_no_outliers[valid_points_mask]
y_valid = clusters_dbsc_tuned[valid_points_mask]

# Ensure there is more than one cluster for silhouette score calculation
if len(set(y_valid)) > 1:
    silhouette_avg_dbscan_tuned = silhouette_score(X_valid, y_valid)
    print(f"Silhouette Score for Tuned DBSCAN (excluding noise): {silhouette_avg_dbscan_tuned:.4f}")
else:
    print("Cannot compute Silhouette Score: Not enough valid clusters (or only one cluster found excluding noise).")
```

### Visualizing the Dendrogram for Agglomerative Clustering

To determine an appropriate number of clusters for Agglomerative Clustering, we can visualize a dendrogram. The dendrogram shows the hierarchical relationships between clusters. We look for the largest vertical distance that can be traversed without crossing any horizontal lines to suggest the optimal number of clusters.

```python
# Generate the linkage matrix
linked_matrix = linkage(X_pca_no_outliers, method='ward')

plt.figure(figsize=(15, 8))
dendrogram(
    linked_matrix,
    truncate_mode='lastp',
    p=30,
    leaf_rotation=90.,
    leaf_font_size=8.,
    show_contracted=True,
    # Show only the last 'p' merged clusters for better readability
    # color_threshold=10  # You can adjust this threshold to cut the dendrogram
)
plt.title('Dendrogram for Agglomerative Clustering (PCA-Reduced Data)')
plt.xlabel('Sample Index or Cluster Size')
plt.ylabel('Distance')
plt.grid(True)
plt.show()
```

From the dendrogram, we can observe distinct clusters. A cut at a certain distance would suggest a number of clusters. Visually, 3 or 4 clusters seem like reasonable choices based on the vertical lines that could be drawn without crossing horizontal lines, similar to the K-Means analysis.

### Applying Agglomerative Clustering with K=3

Based on the dendrogram and consistency with previous methods, let's proceed with 3 clusters.

```python
# Apply Agglomerative Clustering with 3 clusters
agg_clustering_k3 = AgglomerativeClustering(n_clusters=3, linkage='ward')
clusters_agg_k3 = agg_clustering_k3.fit_predict(X_pca_no_outliers)

# Visualize the clusters using the first two principal components
plt.figure(figsize=(10, 8))
for i in range(3):
    plt.scatter(X_pca_no_outliers[clusters_agg_k3 == i, 0], X_pca_no_outliers[clusters_agg_k3 == i, 1], s=100, label=f'Cluster {i+1}')

plt.title('Agglomerative Clustering (K=3) on First Two Principal Components (Outliers Removed)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_avg_agg = silhouette_score(X_pca_no_outliers, clusters_agg_k3)

print(f"Silhouette Score for Agglomerative Clustering (K=3): {silhouette_avg_agg:.4f}")
```

### K-Medoids Clustering

K-Medoids is a clustering algorithm related to K-Means, but instead of using the mean of the points in a cluster as its center, it uses one of the actual data points as the medoid. This makes K-Medoids more robust to noise and outliers compared to K-Means.

We will use the `KMedoids` implementation from `sklearn_extra.cluster`.

```python
# Install scikit-learn-extra if not already installed
!pip install scikit-learn-extra
```

```python
# First, uninstall current numpy and install a compatible version
!pip uninstall -y numpy
!pip install numpy==1.26.4

# NOTE: After installing the older numpy, you MUST restart the runtime (Runtime -> Restart runtime)!
# Then, you can re-run this cell and the subsequent K-Medoids evaluation cell.

from sklearn_extra.cluster import KMedoids
import matplotlib.pyplot as plt

# Apply K-Medoids on the PCA-transformed data (X_pca_no_outliers) with K=3
kmedoids_pca_k3 = KMedoids(n_clusters=3, random_state=42)
y_pca_kmedoids_k3 = kmedoids_pca_k3.fit_predict(X_pca_no_outliers)

# Visualize the clusters using the first two principal components
plt.figure(figsize=(10, 8))
for i in range(3):
    plt.scatter(X_pca_no_outliers[y_pca_kmedoids_k3 == i, 0], X_pca_no_outliers[y_pca_kmedoids_k3 == i, 1], s=100, label=f'Cluster {i+1}')

# Plot the cluster medoids
plt.scatter(kmedoids_pca_k3.cluster_centers_[:, 0], kmedoids_pca_k3.cluster_centers_[:, 1],
            s=300, c='black', marker='X', label='Medoids')

plt.title('K-Medoids Clustering (K=3) on First Two Principal Components')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid(True)
plt.show()
```

### Evaluating K-Medoids Clustering Performance (Silhouette Score)

```python
from sklearn.metrics import silhouette_score

silhouette_avg_kmedoids = silhouette_score(X_pca_no_outliers, y_pca_kmedoids_k3)

print(f"Silhouette Score for K-Medoids (K=3): {silhouette_avg_kmedoids:.4f}")
```
