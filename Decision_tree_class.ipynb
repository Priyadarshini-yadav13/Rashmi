Step 4: Segregating the data into input and output

```python
x=df.data
```

```python
x
```

```python
y=df.target
```

```python
y
```

Step5: split the dataset into training and testing data

```python
x_train,x_test,y_train,y_test=train_test_split( x,y,test_size=.2,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape,y_train.shape,y_test.shape
```

Step6:Workin on Decision tree

Step7:Finding the optimal depth

```python
accuracy=[]
for i in range(1, 11):
  print("depth : ", i)
  model=DecisionTreeClassifier(max_depth=i)
  model.fit(x_train, y_train) # training
  y_pred=model.predict(x_test) # testing
  print("accuracy : ", accuracy_score(y_test, y_pred))
  accuracy.append(accuracy_score(y_test, y_pred))
```

```python
accuracy
```

```python
plt.plot(range(1, 11), accuracy)
```

Based on the visualisation tha max depth=3

```python
#Loading the model
```

```python
model=DecisionTreeClassifier(max_depth=3,criterion='entropy')
```
