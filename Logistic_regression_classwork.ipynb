Step 1:importing important libraries

```python
#basic libraries
import numpy as np
import pandas as pd
#visualisation
import matplotlib.pyplot as plt
import seaborn as sns
#Model and preprocessing tools
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

Step2:loading the dataset

```python
iris=load_iris()
iris
```

```python
iris.data
```

```python
df=pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
df
```

```python
df['species'] = iris.target
```

```python
df
```

#summerising the dataset

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

#Preprocessing the dataset


```python
df.isnull().sum()
```

```python
df.corr()
```

```python
sns.heatmap(df.corr(), annot=True, cmap='Blues')
```

```python
#segregating the dataset into input and output
```

```python
x = df[iris.feature_names]
```

```python
Y = df['species']
```

```python
x
```

```python
Y
```

Performin feature scaling

```python
scaler=StandardScaler() # It is a model which will make the data in the same scale to avoid bias
```

```python
x_scaler=scaler.fit_transform(x) # Apply this function so that the machine will not be biased towards any particular values. General tendency would be to be biased towards larger values and to avoid the same we do this
```

#

```python
#Splitting the dataset into train and test
```

```python
x_train,x_test,Y_train,Y_test=train_test_split(x_scaler,Y,test_size=.2,random_state=42)
print("dataset split successful")
print(f"x_train shape: {x_train.shape}")
print(f"x_test shape: {x_test.shape}")
```

#Loading the model

```python
model = LogisticRegression()
```

Step#9Training the model

```python
model.fit(x_train,Y_train)
```

Step#9 Testing the model

```python
Y_pred=model.predict(x_test)
```

```python
Y_pred
```

Step#10 Calculating the accuracy of the trained model

```python
accuracy_score(Y_test,Y_pred)
```

```python
confusion_matrix(Y_test,Y_pred)
```

```python
sns.heatmap(confusion_matrix(Y_test,Y_pred),annot=True,cmap='Blues')
```

```python
print(classification_report(Y_test,Y_pred))
```
