```python
## segregating the data into dependent and independent variables
X=df.iloc[:,0:13]
y=df.iloc[:,13]
```

```python
X
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Display the first few rows of the scaled data as a DataFrame for better readability
display(pd.DataFrame(X_scaled, columns=X.columns).head())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y = le.fit_transform(y)

# Display the first few elements of the converted y
print(y[:5])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
```

```python
model=Sequential()
model.add(Dense(128,activation='relu',input_dim=X_train.shape[1]))
model.add(Dropout(0.2))
model.add(Dense(64,activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(32,activation='relu'))
model.add(Dropout(0.2))
model.add(Dense(1,activation='sigmoid'))
```

```python
X_train.shape[1]
```

```python
from tensorflow.keras.optimizers import Adam

model.compile(optimizer=Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
history = model.fit(X_train, y_train, epochs=100, batch_size=32, validation_split=0.2)
```

### Training History Visualization

Let's visualize the training and validation accuracy and loss over the epochs to understand the model's learning process and identify potential overfitting.

```python
from tensorflow.keras.callbacks import EarlyStopping

# Define EarlyStopping callback
# Monitor 'val_loss', with a patience of 10 epochs (i.e., wait for 10 epochs if no improvement)
# mode='min' means it will stop when the quantity monitored has stopped decreasing
# restore_best_weights=True will restore model weights from the epoch with the best value of the monitored quantity.
early_stopping = EarlyStopping(monitor='val_loss', patience=10, mode='min', restore_best_weights=True)
```

Now, let's retrain the model with `EarlyStopping` to prevent overfitting.

```python
history = model.fit(X_train, y_train, epochs=100, batch_size=32, validation_split=0.2, callbacks=[early_stopping])
```

Let's plot the training history again to see the effect of Early Stopping.

```python
plt.figure(figsize=(12, 5))

# Plot training & validation accuracy values
plt.subplot(1, 2, 1) # 1 row, 2 columns, 1st plot
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model accuracy with EarlyStopping')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')

# Plot training & validation loss values
plt.subplot(1, 2, 2) # 1 row, 2 columns, 2nd plot
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model loss with EarlyStopping')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')

plt.tight_layout()
plt.show()
```

### Model Evaluation on Test Data

Let's evaluate the model's performance on the independent test dataset (`X_test`, `y_test`) to get an unbiased estimate of its generalization capability.

```python
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f'Test Loss: {loss:.4f}')
print(f'Test Accuracy: {accuracy:.4f}')
```

```python
plt.figure(figsize=(12, 5))

# Plot training & validation accuracy values
plt.subplot(1, 2, 1) # 1 row, 2 columns, 1st plot
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')

# Plot training & validation loss values
plt.subplot(1, 2, 2) # 1 row, 2 columns, 2nd plot
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')

plt.tight_layout()
plt.show()
```

```python
##predicting on test data
y_pred_proba = model.predict(X_test)
y_pred = (y_pred_proba > 0.5).astype(int)

print("First 10 predicted probabilities:")
print(y_pred_proba[:10])
print("\nFirst 10 predicted class labels:")
print(y_pred[:10])
```

```python
# Map 0 to 'Absence' and 1 to 'Presence'
predicted_labels_str = np.where(y_pred == 1, 'Presence', 'Absence')

print("First 10 predicted class labels (mapped):")
print(predicted_labels_str[:10].flatten())
```

### Confusion Matrix

Let's calculate and visualize the confusion matrix to evaluate the classification performance of the model. This matrix shows the number of correct and incorrect predictions made by the classifier compared to the actual outcomes.

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate the confusion matrix
cm = confusion_matrix(y_test, y_pred)

# Get the class names for better labels
class_names = ['Absence', 'Presence']

# Plot the confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title('Confusion Matrix')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```

### Interpreting the Confusion Matrix

-   **True Positives (TP):** The model correctly predicted 'Presence' (heart disease) when it was actually 'Presence'.
-   **True Negatives (TN):** The model correctly predicted 'Absence' (no heart disease) when it was actually 'Absence'.
-   **False Positives (FP):** The model incorrectly predicted 'Presence' when it was actually 'Absence' (Type I error).
-   **False Negatives (FN):** The model incorrectly predicted 'Absence' when it was actually 'Presence' (Type II error).

### Classification Report

Let's generate a classification report to get a more comprehensive evaluation of the model's performance, including precision, recall, and F1-score for each class.

```python
from sklearn.metrics import classification_report

# Generate the classification report
report = classification_report(y_test, y_pred, target_names=['Absence', 'Presence'])

print("Classification Report:")
print(report)
```
