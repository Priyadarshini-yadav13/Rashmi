```python
df.groupby(['Gender'])['Amount'].sum().sort_values(ascending=False)
```

```python
sales_gen=df.groupby(['Gender'])['Amount'].sum().sort_values(ascending=False).reset_index()
print("Type of sales_gen:", type(sales_gen))
print("Head of sales_gen:", sales_gen.head())
sns.barplot(x='Gender',y='Amount',data=sales_gen)
```

From above graph we can see that most of the buyers are female and female are having more purchase power.

AGE

The above Graph shows that most of the buyers lie in the age group of 26-35.

Marital status

```python
ax=sns.countplot(x='Marital_Status',data=df,hue='Gender')
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
marital_amount=df.groupby(['Marital_Status', 'Gender'])['Amount'].sum().sort_values(ascending=False).reset_index().head(10)
sns.set(rc={'figure.figsize':(15,5)})
ax=sns.barplot(x='Marital_Status',y='Amount',data=marital_amount,hue='Gender')
for bars in ax.containers:
  ax.bar_label(bars)
```

 the above quote says that married females spend more amount on the purchase.

Occupation

```python
sns.set(rc={'figure.figsize':(15,5)})
ax4=sns.countplot(x='Occupation',data=df)
for bars in ax4.containers:
  ax4.bar_label(bars)
```

The above analysis says that the IT sector people order the most.

```python
sns.set(rc={'figure.figsize':(15,5)})
occupation_amount=df.groupby('Occupation')['Amount'].sum().sort_values(ascending=False).reset_index().head(15)
ax5=sns.barplot(x='Occupation',y='Amount',data=occupation_amount)
for bars in ax5.containers:
  ax5.bar_label(bars)
```

From the above graph, it is said that, first, we can see that most of the buyers are working in IT, then health care, and then aviation.

Product Category

```python
sns.set(rc={'figure.figsize':(25,5)})
ax6=sns.countplot(x='Product_Category',data=df)
for bars in ax6.containers:
  ax6.bar_label(bars)
```

The above analysis says that most of the products which are ordered are clothing and apparel.

```python
df2=df.groupby('Product_Category')['Amount'].sum().sort_values(ascending=False).reset_index().head(10)
sns.set(rc={'figure.figsize':(15,5)})
ax6=sns.barplot(x='Product_Category',y='Amount',data=df2)
for bars in ax6.containers:
  ax6.bar_label(bars)
```

Most of the amount for product category food is used.

```python
df2=df.groupby('Product_ID')['Orders'].sum().sort_values(ascending=False).reset_index().head(10)
sns.set(rc={'figure.figsize':(15,5)})
ax7=sns.barplot(x='Product_ID',y='Orders',data=df2)
for bars in ax7.containers:
  ax7.bar_label(bars)
```

Married women aged 26-35 from UP, Maharashtra, and Karnataka working in IT, healthcare, and aviation are more likely to buy products from the food, clothing, and electronics category.
