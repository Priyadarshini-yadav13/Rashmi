```python
## importing more libraries
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense ## For creating neurons at all levels
from tensorflow.keras.layers import Dropout ##for preventing overfitting
from tensorflow.keras.layers import Conv2D ##For feature extracting
from tensorflow.keras.layers import MaxPooling2D ## For downsampling the data
from tensorflow.keras.layers import ReLU, Softmax, Flatten
from tensorflow.keras.utils import to_categorical
```

## Pre-process the data.

```python
# Separate features (X) and labels (y)
X_train = df_train.drop('label', axis=1).values
y_train = df_train['label'].values

X_test = df_test.drop('label', axis=1).values
y_test = df_test['label'].values
```

```python
# Display one sample image from the training data
plt.imshow(X_train[0].reshape(28, 28), cmap='gray')
plt.title(f"Label: {y_train[0]}")
plt.axis('off')
plt.show()
```

Let's inspect the unique class labels present in the `y_test` dataset. This will show us the different categories our model is trying to predict.

```python
print(f"Unique class labels in y_test: {np.unique(y_test)}")
```

The Fashion MNIST dataset has 10 classes, which are mapped to numerical labels as follows:

*   0: T-shirt/top

```python
# Define a mapping from numerical labels to class names
label_names = {
    0: 'T-shirt/top',
    1: 'Trouser',
    2: 'Pullover',
    3: 'Dress',
    4: 'Coat',
    5: 'Sandal',
    6: 'Shirt',
    7: 'Sneaker',
    8: 'Bag',
    9: 'Ankle boot'
}

# Display the unique class names present in y_test
unique_labels_in_test = np.unique(y_test)
class_names_in_test = [label_names[label] for label in unique_labels_in_test]
print(f"Class names present in y_test: {class_names_in_test}")
```

```python
# Normalize the data
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
```

```python
# Reshape the data for CNN input (28x28 grayscale images)
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)
```

```python
# One-hot encode the labels
num_classes = 10 # Fashion MNIST has 10 classes
y_train_encoded = to_categorical(y_train, num_classes)
y_test_encoded = to_categorical(y_test, num_classes)
```

The data has been pre-processed and is now ready for building the CNN model.

```python
# Build the CNN model
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5), # Add dropout for regularization
    Dense(num_classes, activation='softmax')
])

# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

model.summary()
```

The CNN model has been constructed and compiled. The `model.summary()` output above provides an overview of the layers and parameters in the network. We can now proceed to train the model.

The CNN model has been constructed and compiled. The `model.summary()` output above provides an overview of the layers and parameters in the network. We can now proceed to train the model.

```python
# Train the model
history = model.fit(
    X_train, y_train_encoded,
    epochs=10, # You can adjust the number of epochs
    batch_size=32, # You can adjust the batch size
    validation_data=(X_test, y_test_encoded),verbose=1
)

print("Model training complete.")
```

The model has finished training. The `history` object contains the training and validation metrics over the epochs. Next, we can evaluate the model's performance and visualize the training history.

```python
acc_cnn=model.evaluate(X_test,y_test_encoded,verbose=1)[1]
```

```python
acc_cnn
```

Let's visualize the training and validation accuracy and loss over the epochs to assess the model's learning progress and identify potential overfitting or underfitting.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 6))

# Plot Training and Validation Accuracy
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# Plot Training and Validation Loss
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()
```

The above training and validation accuracy curve says that as our epoch increases, accuracy is increasing, which is appropriate according to the rule that accuracy should always be in increasing order.
```python
from sklearn.metrics import classification_report
import numpy as np

# Make predictions on the test set
y_pred_probs = model.predict(X_test)
y_pred = np.argmax(y_pred_probs, axis=1)

# Convert y_test_encoded back to original labels for comparison
y_test_labels = np.argmax(y_test_encoded, axis=1)

# Generate and print the classification report
print("Classification Report:")
print(classification_report(y_test_labels, y_pred, target_names=[label_names[i] for i in range(num_classes)]))
```

### Fine-tuning the CNN Model

To fine-tune the CNN model for better performance, we can consider several strategies:

1.  **Increase Model Complexity**: Add more convolutional layers, increase the number of filters in existing layers, or use larger kernel sizes (with caution).
2.  **Adjust Hyperparameters**: Experiment with different learning rates, batch sizes, and the number of epochs.
3.  **Regularization**: Introduce more `Dropout` layers, or add L1/L2 regularization to convolutional or dense layers to combat overfitting.
4.  **Activation Functions**: Try different activation functions like `LeakyReLU` or `ELU`.
5.  **Data Augmentation**: Apply transformations like rotations, shifts, and flips to the training images to increase the diversity of the dataset and make the model more robust.
6.  **Optimizer**: Experiment with other optimizers like SGD with momentum or RMSprop.

Let's try increasing the model complexity by adding another `Conv2D` layer and slightly increasing the filter count.

```python
# Build a more complex CNN model (Fine-tuned version)
model_tuned = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'), # Added an additional Conv2D layer
    MaxPooling2D((2, 2)), # Added a MaxPooling layer for the new Conv2D layer
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(num_classes, activation='softmax')
])

# Compile the tuned model
model_tuned.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("Fine-tuned model summary:")
model_tuned.summary()
```

```python
# Train the fine-tuned model
history_tuned = model_tuned.fit(
    X_train, y_train_encoded,
    epochs=10, # Keep the number of epochs consistent for comparison
    batch_size=32, # Keep the batch size consistent
    validation_data=(X_test, y_test_encoded),
    verbose=1
)

print("Fine-tuned model training complete.")
```

```python
# Evaluate the fine-tuned model
acc_cnn_tuned = model_tuned.evaluate(X_test, y_test_encoded, verbose=1)[1]
print(f"Fine-tuned CNN model test accuracy: {acc_cnn_tuned:.4f}")
```

5.  **Reduce Model Complexity**: Counter-intuitively, reducing some of the added complexity might be beneficial, as the initial model performed better. We could try a model architecture between the initial and the current fine-tuned one.
6.  **Hyperparameter Tuning**: Revisit learning rate and batch size. A smaller learning rate might allow the model to converge more smoothly without overshooting the optimal weights.

### Implementing Regularization: Increased Dropout and Early Stopping

To combat the observed overfitting, we will create a new model (`model_regularized`) with the following changes:

1.  **Increased Dropout**: We'll increase the dropout rate in the existing `Dropout` layer from 0.5 to 0.7.
2.  **Early Stopping**: We'll use the `EarlyStopping` callback during training to monitor the validation loss and stop training if it doesn't improve for a certain number of epochs (patience).

```python
from tensorflow.keras.callbacks import EarlyStopping

# Define Early Stopping callback
# Monitor 'val_loss', with a patience of 5 epochs (stop if val_loss doesn't improve for 5 consecutive epochs)
early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

# Build a regularized CNN model
model_regularized = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.7), # Increased dropout rate
    Dense(num_classes, activation='softmax')
])

# Compile the regularized model
model_regularized.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("Regularized model summary:")
model_regularized.summary()
```

```python
# Train the regularized model with Early Stopping
history_regularized = model_regularized.fit(
    X_train, y_train_encoded,
    epochs=50, # Set a higher number of epochs as Early Stopping will manage stopping
    batch_size=32,
    validation_data=(X_test, y_test_encoded),
    callbacks=[early_stopping], # Add the EarlyStopping callback
    verbose=1
)

print("Regularized model training complete.")
```

```python
# Evaluate the regularized model
acc_cnn_regularized = model_regularized.evaluate(X_test, y_test_encoded, verbose=1)[1]
print(f"Regularized CNN model test accuracy: {acc_cnn_regularized:.4f}")
```

### Evaluating the Regularized Model

Now let's evaluate the `model_regularized` by generating its classification report and plotting its training and validation curves to see if overfitting has been mitigated and if performance has improved.

```python
# Make predictions on the test set with the regularized model
y_pred_probs_regularized = model_regularized.predict(X_test)
y_pred_regularized = np.argmax(y_pred_probs_regularized, axis=1)

# Generate and print the classification report for the regularized model
print("Classification Report for Regularized Model:")
print(classification_report(y_test_labels, y_pred_regularized, target_names=[label_names[i] for i in range(num_classes)]))
```

```python
plt.figure(figsize=(15, 6))

# Plot Training and Validation Accuracy for regularized model
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
plt.plot(history_regularized.history['accuracy'], label='Training Accuracy')
plt.plot(history_regularized.history['val_accuracy'], label='Validation Accuracy')
plt.title('Regularized Model: Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# Plot Training and Validation Loss for regularized model
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
plt.plot(history_regularized.history['loss'], label='Training Loss')
plt.plot(history_regularized.history['val_loss'], label='Validation Loss')
plt.title('Regularized Model: Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

### Implementing Data Augmentation

Data augmentation helps to increase the diversity of the training data, which in turn can lead to more robust models that generalize better to unseen data. We will use `ImageDataGenerator` from Keras to apply random transformations to our training images.

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Create an ImageDataGenerator object with augmentation parameters
datagen = ImageDataGenerator(
    rotation_range=10,        # randomly rotate images in the range (degrees, 0 to 180)
    zoom_range=0.1,           # randomly zoom image
    width_shift_range=0.1,    # randomly shift images horizontally (fraction of total width)
    height_shift_range=0.1,   # randomly shift images vertically (fraction of total height)
    shear_range=0.1,          # randomly apply shearing transformations
    horizontal_flip=True,     # randomly flip images horizontally
    fill_mode='nearest'       # strategy for filling in newly created pixels
)

# Fit the data generator on the training data
datagen.fit(X_train)

print("ImageDataGenerator configured and fitted.")
```

### Building and Training a Model with Data Augmentation

Now, let's build a new model (`model_augmented`) with the same architecture as our `model_regularized` but train it using the augmented data generated by `datagen.flow()`. We'll also keep the `EarlyStopping` callback to prevent overfitting.

```python
# Re-define Early Stopping callback (if not already in scope)
from tensorflow.keras.callbacks import EarlyStopping
early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

# Build a new CNN model (same architecture as regularized model)
model_augmented = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.7), # Increased dropout rate
    Dense(num_classes, activation='softmax')
])

# Compile the augmented model
model_augmented.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("Augmented model summary:")
model_augmented.summary()
```

```python
# Train the model using augmented data from the generator
history_augmented = model_augmented.fit(
    datagen.flow(X_train, y_train_encoded, batch_size=32), # Use generator for training data
    epochs=50, # Set a higher number of epochs as Early Stopping will manage stopping
    validation_data=(X_test, y_test_encoded),
    callbacks=[early_stopping],
    verbose=1
)

print("Augmented model training complete.")
```

### Evaluating the Augmented Model

Let's evaluate the `model_augmented` to see if data augmentation has further improved generalization and overall performance. We will check its test accuracy, classification report, and visualize its training/validation curves.

```python
# Evaluate the augmented model
acc_cnn_augmented = model_augmented.evaluate(X_test, y_test_encoded, verbose=1)[1]
print(f"Augmented CNN model test accuracy: {acc_cnn_augmented:.4f}")
```
