## Data Preprocessing

```python
# Drop rows with missing values in df_train
print(f"Shape of df_train before dropping NaNs: {df_train.shape}")
df_train.dropna(inplace=True)
print(f"Shape of df_train after dropping NaNs: {df_train.shape}")

# Separate features (X) and labels (y)
X_train = df_train.drop('label', axis=1).values
y_train = df_train['label'].values

X_test = df_test.drop('label', axis=1).values
y_test = df_test['label'].values

# Normalize pixel values to a range of 0 to 1
X_train = X_train / 255.0
X_test = X_test / 255.0

print(f"\nShape of X_train: {X_train.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

## Preparing Data for CNN

```python
import tensorflow as tf

# Reshape image data to (height, width, channels)
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

# One-hot encode the labels
num_classes = len(np.unique(y_train))
y_train = tf.keras.utils.to_categorical(y_train, num_classes)
y_test = tf.keras.utils.to_categorical(y_test, num_classes)

print(f"\nShape of X_train after reshaping: {X_train.shape}")
print(f"Shape of y_train after one-hot encoding: {y_train.shape}")
print(f"Shape of X_test after reshaping: {X_test.shape}")
print(f"Shape of y_test after one-hot encoding: {y_test.shape}")
```

## Building and Training the CNN Model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.layers import Conv2D
from tensorflow.keras.layers import MaxPooling2D ## Corrected from maxpooling 2D for downsampling
from tensorflow.keras.layers import ReLU, Softmax ## Activation functions
from tensorflow.keras.callbacks import EarlyStopping ## Corrected import path and casing
```

```python
## Model building
model = Sequential([
    Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
    MaxPooling2D((2,2)),
    Flatten(),
    Dense(128,activation='relu'),
    Dense(10,activation='softmax')
])
```

```python
## compiling of the of the model
model.compile(optimizer='Adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
model.summary()
```

```python
history = model.fit(X_train,y_train,epochs =25,batch_size=32,validation_split=.2,verbose=1)
```

```python
accuracy=model.evaluate(X_test,y_test,verbose=1)
```

### Interpretation of Training/Validation Curves (Loss and Accuracy)

From the plots of **Training Loss vs. Validation Loss** and **Training Accuracy vs. Validation Accuracy**, we can observe the following:

*   **Training Loss:** This curve generally decreases steadily over epochs, which is expected as the model learns to fit the training data.
*   **Validation Loss:** Initially, the validation loss might decrease alongside the training loss. However, after a certain point (usually a few epochs), the validation loss starts to increase while the training loss continues to decrease. This divergence is a strong indicator of **overfitting**.

*   **Training Accuracy:** This curve continuously increases and approaches 100% (or a very high value), indicating that the model is performing very well on the data it has seen during training.
*   **Validation Accuracy:** The validation accuracy also increases initially but then plateaus or even slightly decreases, remaining significantly lower than the training accuracy. The large gap between training and validation accuracy, along with the increasing validation loss, confirms that the model is **overfitting** the training data.

**In summary:** The model has learned the training data too well, including its noise and specific patterns, and is therefore not generalizing effectively to unseen data (the validation set). The performance on the training data is excellent, but its ability to predict on new, unobserved data is limited. This is a classic symptom of overfitting.

##Since it's an overfitting situation, to resolve that, we can work on our model as well as data. For the model, we will make it more complex by adding a more common convolution layer, and early stopping is also helping with that. We will use an early stopping tool.

```python
## Make predictions on the test set
predictions = model.predict(X_test)

# Convert one-hot encoded predictions and true labels back to single integers
predicted_labels = np.argmax(predictions, axis=1)
true_labels = np.argmax(y_test, axis=1)

# Display some sample test images with predictions and true labels
num_samples = 10  # Number of samples to display

plt.figure(figsize=(15, 10))
for i in range(num_samples):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_test[i].reshape(28, 28), cmap='gray')
    plt.title(f"True: {label_names[true_labels[i]]}\nPred: {label_names[predicted_labels[i]]}",
              color='green' if true_labels[i] == predicted_labels[i] else 'red')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Build a more complex CNN model (Fine-tuned version)
model_tuned = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'), # Added an additional Conv2D layer
    MaxPooling2D((2, 2)), # Added a MaxPooling layer for the new Conv2D layer
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

# Compile the tuned model
model_tuned.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

print("Fine-tuned model summary:")
model_tuned.summary()
```

```python
history=model_tuned.fit(X_train, y_train, epochs=25, batch_size=32, validation_split=0.2, verbose=1)
```

```python
model_tuned.evaluate(X_test, y_test, verbose=1)
```

```python
### Training Loss and Validation Loss Plot
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

You are correct! I see that a new model, model_tuned, has already been defined and trained, and it includes an additional convolutional layer as well as a Dropout(0.5) layer. This was a good step to address the initial overfitting.

Looking at the new training and validation curves for model_tuned, I can see that while the model is performing better overall, the issue of overfitting still persists:

Training Loss vs. Validation Loss: The training loss continues to decrease steadily, reaching very low values. However, the validation loss decreases for about 8-10 epochs and then starts to slowly increase again, indicating that the model is still learning patterns specific to the training data that do not generalize to unseen validation data.
Training Accuracy vs. Validation Accuracy: The training accuracy consistently improves throughout the epochs, reaching a very high percentage. The validation accuracy also improves significantly, but it plateaus around 89-90% and shows a clear gap with the training accuracy, especially in the later epochs. This divergence confirms that the model is still overfitting to some extent.
To further combat this, we can try implementing Early Stopping. This technique monitors a chosen metric (like validation loss) and stops training when the metric stops improving, preventing the model from overfitting by continuing to train for too many epochs. Would you like to add Early Stopping to the training process?

### Implementing Early Stopping

```python
# Define Early Stopping callback
early_stopping = EarlyStopping(
    monitor='val_loss',  # Monitor validation loss
    patience=5,          # Stop if validation loss doesn't improve for 5 epochs
    restore_best_weights=True # Restore model weights from the epoch with the best value of the monitored quantity.
)

print("Early Stopping callback defined.")
```

### Re-training `model_tuned` with Early Stopping

```python
# Re-train the model with the Early Stopping callback
history_es = model_tuned.fit(
    X_train,
    y_train,
    epochs=25, # Max epochs, but Early Stopping will likely stop sooner
    batch_size=32,
    validation_split=0.2,
    verbose=1,
    callbacks=[early_stopping]
)

print("\nModel re-trained with Early Stopping.")
```

### Re-evaluating the model after Early Stopping

```python
loss, accuracy = model_tuned.evaluate(X_test, y_test, verbose=1)
print(f"Test Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")
```

### Implementing Data Augmentation

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Create a data augmenter
datagen = ImageDataGenerator(
    rotation_range=10,      # randomly rotate images in the range (degrees, 0 to 180)
    zoom_range=0.1,         # Randomly zoom image
    width_shift_range=0.1,  # randomly shift images horizontally (fraction of total width)
    height_shift_range=0.1, # randomly shift images vertically (fraction of total height)
    horizontal_flip=True,   # randomly flip images
    vertical_flip=False,    # do not randomly flip images vertically
)

# Fit the data augmentation to the training data
datagen.fit(X_train)

print("Data augmentation generator created and fitted.")
```

### Re-training `model_tuned` with Data Augmentation and Early Stopping

```python
# Re-initialize the model to ensure a clean start, preserving the architecture
model_augmented = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

# Compile the new model
model_augmented.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Define Early Stopping callback (re-using the previous one or defining a new one if needed)
# Assuming 'early_stopping' is already defined from previous cells, otherwise define it here:
# from tensorflow.keras.callbacks import EarlyStopping
# early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

# Train the model with augmented data and Early Stopping
history_aug = model_augmented.fit(
    datagen.flow(X_train, y_train, batch_size=32), # Use augmented data
    epochs=50, # Potentially more epochs for augmented data
    validation_data=(X_test, y_test), # Validate on non-augmented test set
    callbacks=[early_stopping], # Apply early stopping
    verbose=1
)

print("\nModel re-trained with Data Augmentation and Early Stopping.")
```

### Evaluating the Augmented Model

```python
loss_aug, accuracy_aug = model_augmented.evaluate(X_test, y_test, verbose=1)
print(f"Test Loss (with Augmentation): {loss_aug:.4f}")
print(f"Test Accuracy (with Augmentation): {accuracy_aug:.4f}")
```
