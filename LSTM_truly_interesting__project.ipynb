```python
Data="""Melissa sneaked behind the teacher as he wrote an equation on the board.
She sat on her seat, trying hard not to make a noise.
The teacher turned around to explain what he had written on the board.
He turned around and looked confused. “Melissa, I didn’t see you.
Were you not there in the morning?” asked teacher.
I honestly think you need new glasses at this point”, joked Melissa.
The whole room started to laugh.Teacher ignored it and got back to teaching.
Melissa was relieved she didn’t get caught sneaking in.
She did not want anyone knowing the reason she was late.
She partied all night long with her friend Meleena who was absent that day, came home at 3 and slept really late.
She barely had enough time to eat her breakfast in the morning as she woke up so late.
Meleena was her best friend and helped her with all her problems, Melissa missed her at school.
The day went by with boring classes and high school drama Melissa did not want to be a part of.
She went home and cooked some lasagna.
She cooked for herself as her parents weren’t home.
They had gone to Paris for celebrating their 20th anniversary.
They refused to take Melissa, because they thought she would ruin their vacation. Melissa wasn’t mad at them, she agreed.
After finishing her lasagna, she took a warm shower, meditated, took her medication and went to sleep.
Her alarm ringed when it was 4 in the morning.
She woke up with her brain barely functioning and her body aching.
First, she drank a warm glass of water.
Then, she took a shower, brushed her teeth, washed her face and wore comfortable clothing."""
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts([Data])
```

```python
for sentence in Data.split('.'):
    print(sentence)
```

```python
print(tokenizer.texts_to_sequences([Data]))
```

```python
for sentence in Data.split('.'):
    print(tokenizer.texts_to_sequences([sentence])[0])
```

```python
input_sequences=[]
for sentence in Data.split('.'):
  tokenized_sentence=tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
input_sequences
```

```python
max_len=max([len(x) for x in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences=pad_sequences(input_sequences,maxlen=max_len,padding='pre')
```

```python
padded_input_sequences
```

```python
X=padded_input_sequences[:,:-1]
```

```python
y=padded_input_sequences[:,-1]
```

```python
X
```

```python
y
```

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y,num_classes=len(tokenizer.word_index)+1)
```

```python
y.shape
```

```python
y
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense
```

```python
model=Sequential()
model.add(Embedding(len(tokenizer.word_index)+1,100,input_length=max_len-1))
model.add(LSTM(150))
model.add(Dense(len(tokenizer.word_index)+1,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.summary()
```

```python
# Explicitly build the model with the input shape of X to force parameter initialization
model.build(input_shape=(None, X.shape[1]))
model.summary()
```

```python
model.fit(X,y,epochs=100)
```

```python
text="melissa"
#tokenize
token_text=tokenizer.texts_to_sequences([text])[0]
##padding
padded_token_text=pad_sequences([token_text],maxlen=max_len-1,padding='pre')
print(padded_token_text)
##predict
pos=np.argmax(model.predict(padded_token_text))
for word,index in tokenizer.word_index.items():
  if index==pos:
    print(word)
```

```python
## for 12 words predictions
text="melissa"
for i in range(12):
  #tokenize
  token_text=tokenizer.texts_to_sequences([text])[0]
  ##padding
  padded_token_text=pad_sequences([token_text],maxlen=max_len-1,padding='pre')
  print(padded_token_text)
  ##predict
  pos=np.argmax(model.predict(padded_token_text, verbose=0))
  for word,index in tokenizer.word_index.items():
    if index==pos:
      text=text+" "+word
      print(text)
```
